```python
import numpy as np

def calculate_mutual_information(counts_table):
    """
    Calculates mutual information from a 2x2 contingency table.
    counts_table = [[N_11, N_10], [N_01, N_00]]
    """
    N = np.sum(counts_table)

    # Joint probabilities
    P_xy = counts_table / N

    # Marginal probabilities
    P_x = np.sum(counts_table, axis=1) / N # P(S=1), P(S=0)
    P_y = np.sum(counts_table, axis=0) / N # P(K=1), P(K=0)

    mi = 0.0
    for i in range(2):
        for j in range(2):
            if P_xy[i, j] > 0:
                mi += P_xy[i, j] * np.log2(P_xy[i, j] / (P_x[i] * P_y[j]))
    return mi
```

```python
# Counts for "prize"
counts_prize = np.array([[130, 15],
                         [1200, 13000]])
mi_prize = calculate_mutual_information(counts_prize)
print(f"Mutual Information for 'prize': {mi_prize:.5f} bits")

# Counts for "hello"
counts_hello = np.array([[160, 25],
                         [13000, 7500]])
mi_hello = calculate_mutual_information(counts_hello)
print(f"Mutual Information for 'hello': {mi_hello:.5f} bits")

# Output from Python:
# Mutual Information for 'prize': 0.014807 bits
# Mutual Information for 'hello': 0.000303 bits
```
