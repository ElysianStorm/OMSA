```python
# Cell 1: Introduction and Data Loading
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import io

print("### Introduction to PCA and Data Loading")
print("This notebook demonstrates Principal Component Analysis (PCA) on the 'food-consumption.csv' dataset.")
print("PCA is a dimensionality reduction technique that transforms a set of correlated variables into a set of linearly uncorrelated variables called principal components.")
print("The first few principal components capture the most variance in the data, allowing us to represent high-dimensional data in a lower-dimensional space while retaining most of the information.")

# Load the dataset
df = pd.read_csv("./data/food-consumption.csv")

print("\n--- Raw Data ---")
print("We begin by loading the 'food-consumption.csv' dataset into a pandas DataFrame.")
print("This dataset contains consumption data for 20 different food items across 16 European countries.")
print("\nFirst 5 rows of the dataset:")
print(df.head())
print("\nBasic information about the dataset:")
print(df.info())
```

```python
# Cell 2: Part 1 - Data Matrix Setup for Countries PCA
print("### Part 1: PCA on Countries (treating food consumption as features)")
print("\nFor this part, we consider each country's food consumption as a 'feature vector'.")
print("This means we are looking for patterns among countries based on their consumption habits.")

print("\n#### Data Matrix Setup")
print("In this setup, each row of our data matrix represents a unique 'Country', and each column represents a 'Food Item' (feature).")
print("The values in the matrix correspond to the consumption levels of each food item in each country.")
print("The original dataset has 'Country' as a column, which is an identifier. For PCA, we will set 'Country' as the index.")

df_pca_countries = df.set_index('Country')

print("\nData matrix `df_pca_countries` after setting 'Country' as index (first 5 rows):")
print(df_pca_countries.head())
print(f"The shape of this data matrix is: {df_pca_countries.shape}")
print("This means we have 16 observations (countries) and 20 features (food items).")
```

```python
# Cell 3: Part 1 - Mean-centering the Data for Countries PCA
print("#### Mean-centering the Data")
print("Before performing PCA, it is crucial to mean-center the data. This involves subtracting the mean of each feature (food item) from all its values.")
print("Mean-centering ensures that the principal components are independent of the choice of origin and accurately reflect the variance in the data.")
print("The mean of each column (food item) is calculated and then subtracted from that column.")

mean_consumption_countries = df_pca_countries.mean(axis=0)
data_centered_countries = df_pca_countries - mean_consumption_countries

print("\nMean consumption for each food item (first 5 values):")
print(mean_consumption_countries.head())
print("\nMean-centered data (first 5 rows):")
print(data_centered_countries.head())
print("Notice that the values are now centered around zero for each column.")
```

```python
# Cell 4: Part 1 - Covariance Matrix Calculation for Countries PCA
print("#### Calculating the Covariance Matrix")
print("The covariance matrix is a square matrix that describes the variance within each feature and the covariance between every pair of features.")
print("For $N$ features, the covariance matrix will be of size $N \times N$. In our case, with 20 food items, it will be a $20 \times 20$ matrix.")
print("The diagonal elements represent the variance of each food item, while off-diagonal elements represent the covariance between two food items.")
print("We use `np.cov` with `rowvar=False` because our features are in columns.")

covariance_matrix_countries = np.cov(data_centered_countries, rowvar=False)

print("\nShape of the covariance matrix for countries PCA:", covariance_matrix_countries.shape)
print("First 3x3 block of the covariance matrix:")
print(covariance_matrix_countries[:3, :3])
```

```python
# Cell 5: Part 1 - Eigenvalues and Eigenvectors for Countries PCA
print("#### Computing Eigenvalues and Eigenvectors")
print("Eigenvalues and eigenvectors are fundamental to PCA. Eigenvectors (principal components) define the new directions (axes) in the data space, and eigenvalues indicate the magnitude of variance along these directions.")
print("We compute the eigenvalues and eigenvectors of the covariance matrix. The eigenvector corresponding to the largest eigenvalue is the first principal component, capturing the most variance.")

eigenvalues_countries, eigenvectors_countries = np.linalg.eigh(covariance_matrix_countries)

# Sort eigenvalues and eigenvectors in descending order
sorted_indices_countries = np.argsort(eigenvalues_countries)[::-1]
eigenvalues_countries_sorted = eigenvalues_countries[sorted_indices_countries]
eigenvectors_countries_sorted = eigenvectors_countries[:, sorted_indices_countries]

print("\nSorted Eigenvalues (top 5 for countries PCA):")
print(eigenvalues_countries_sorted[:5])
print("\nThe eigenvalues represent the amount of variance explained by each principal component.")
print("The first few eigenvalues are significantly larger, indicating they capture a large portion of the data's variance.")

print("\nFirst two principal components (eigenvectors):")
principal_components_countries = eigenvectors_countries_sorted[:, :2]
print(principal_components_countries)
print("Each column here is a principal component, which is a vector of weights that combine the original food item consumptions.")
```

```python
# Cell 5: Part 1 - Eigenvalues and Eigenvectors for Countries PCA
print("#### Computing Eigenvalues and Eigenvectors")
print("Eigenvalues and eigenvectors are fundamental to PCA. Eigenvectors (principal components) define the new directions (axes) in the data space, and eigenvalues indicate the magnitude of variance along these directions.")
print("We compute the eigenvalues and eigenvectors of the covariance matrix. The eigenvector corresponding to the largest eigenvalue is the first principal component, capturing the most variance.")

eigenvalues_countries, eigenvectors_countries = np.linalg.eigh(covariance_matrix_countries)

# Sort eigenvalues and eigenvectors in descending order
sorted_indices_countries = np.argsort(eigenvalues_countries)[::-1]
eigenvalues_countries_sorted = eigenvalues_countries[sorted_indices_countries]
eigenvectors_countries_sorted = eigenvectors_countries[:, sorted_indices_countries]

print("\nSorted Eigenvalues (top 5 for countries PCA):")
print(eigenvalues_countries_sorted[:5])
print("\nThe eigenvalues represent the amount of variance explained by each principal component.")
print("The first few eigenvalues are significantly larger, indicating they capture a large portion of the data's variance.")

print("\nFirst two principal components (eigenvectors):")
principal_components_countries = eigenvectors_countries_sorted[:, :2]
print(principal_components_countries)
print("Each column here is a principal component, which is a vector of weights that combine the original food item consumptions.")

# Cell 6: Part 1 - Projecting Data and Scatter Plot for Countries PCA
print("#### Projecting Data onto Principal Components and Scatter Plot")
print("Now, we project the mean-centered data onto the first two principal components.")
print("This transforms our 20-dimensional country data into a 2-dimensional representation, which can be easily visualized.")
print("Each country will now have two coordinates (PC1 and PC2) that represent its position in the reduced space.")

projected_data_countries = np.dot(data_centered_countries, principal_components_countries)

# Create a DataFrame for the projected data with country names
projected_df_countries = pd.DataFrame(projected_data_countries,
                                      columns=['PC1', 'PC2'],
                                      index=df_pca_countries.index)

print("\nProjected data for countries (first 5 rows):")
print(projected_df_countries.head())

print("\nWe will now generate a scatter plot to visualize the countries in this 2D principal component space.")
plt.figure(figsize=(12, 9))
plt.scatter(projected_df_countries['PC1'], projected_df_countries['PC2'], s=100, alpha=0.7)

# Annotate each point with the country name
for i, country in enumerate(projected_df_countries.index):
    plt.annotate(country, (projected_df_countries['PC1'][i], projected_df_countries['PC2'][i]),
                 textcoords="offset points", xytext=(5,5), ha='center')

plt.title('PCA of Countries based on Food Consumption')
plt.xlabel('Principal Component 1 (PC1)')
plt.ylabel('Principal Component 2 (PC2)')
plt.grid(True)
plt.axhline(0, color='grey', linewidth=0.8)
plt.axvline(0, color='grey', linewidth=0.8)
plt.show() # Display the plot in the notebook
```

```python
# Cell 7: Part 1 - Analysis of the Scatter Plot for Countries PCA
print("#### Analysis of the Scatter Plot (Countries PCA)")
print("Observing the scatter plot of countries based on their first two principal components, we can identify several interesting patterns:")
print("- **Clustering:** Countries with similar food consumption patterns tend to cluster together.")
print("  - For example, 'England', 'Ireland', and 'Scotland' (UK countries) are grouped closely, indicating similar dietary habits.")
print("  - 'Portugal' and 'Spain' also appear close, suggesting a common Iberian dietary pattern.")
print("  - Scandinavian countries like 'Sweden', 'Norway', 'Denmark', and 'Finland' form another distinct cluster.")
print("- **Opposing Consumption Patterns:** Countries that are far apart on the plot, especially along the principal axes, tend to have very different consumption profiles.")
print("  - For instance, countries in the bottom-left quadrant (e.g., 'Italy', 'France') seem to have distinct consumption patterns compared to those in the top-right (e.g., 'Germany', 'Holland').")
print("- **Interpretation of Axes (PC1 and PC2):**")
print("  - **PC1 (Horizontal Axis):** This component explains the most variance. Countries on the right side of PC1 might consume more of certain food items that contribute positively to PC1's eigenvector, while countries on the left consume less of those items or more of items contributing negatively.")
print("  - **PC2 (Vertical Axis):** This component explains the second most variance. Countries at the top of PC2 might have higher consumption of certain food items that contribute positively to PC2, and vice versa for countries at the bottom.")
print("  - For a deeper interpretation, one would examine the loadings (eigenvector values) of the principal components to understand which food items contribute most to each PC.")
print("This visualization effectively reduces the complexity of 20 food items to two dimensions, revealing underlying structures and similarities/differences among European countries' food consumption.")
```

```python
# Cell 8: Part 2 - Data Matrix Setup for Food Items PCA
print("### Part 2: PCA on Food Items (treating country consumptions as features)")
print("\nNow, we shift our perspective and perform PCA on the food items themselves, treating the consumption by each country as a feature for that food item.")
print("This approach helps us understand how different food items relate to each other based on their consumption patterns across countries.")

print("\n#### Data Matrix Setup (Transposed)")
print("For this PCA, each row of our data matrix will represent a 'Food Item', and each column will represent a 'Country' (feature).")
print("To achieve this, we need to transpose the original data matrix (excluding the 'Country' identifier column).")

df_pca_food_items = df.set_index('Country').T

print("\nData matrix `df_pca_food_items` after transposing (first 5 rows):")
print(df_pca_food_items.head())
print(f"The shape of this data matrix is: {df_pca_food_items.shape}")
print("This means we now have 20 observations (food items) and 16 features (countries).")
```

```python
# Cell 9: Part 2 - Mean-centering the Data for Food Items PCA
print("#### Mean-centering the Data (Food Items PCA)")
print("Similar to the previous PCA, we mean-center the data. This time, we subtract the mean consumption across all food items for each country (column).")
print("This standardizes the data such that each country's average consumption is zero, allowing us to focus on the relative consumption patterns of food items.")

mean_consumption_food_items = df_pca_food_items.mean(axis=0)
data_centered_food_items = df_pca_food_items - mean_consumption_food_items

print("\nMean consumption for each country (first 5 values):")
print(mean_consumption_food_items.head())
print("\nMean-centered data for food items (first 5 rows):")
print(data_centered_food_items.head())
```

```python
# Cell 10: Part 2 - Covariance Matrix Calculation for Food Items PCA
print("#### Calculating the Covariance Matrix (Food Items PCA)")
print("Here, the covariance matrix will describe the variance within each country's consumption data and the covariance between country consumption patterns.")
print("Since we have 16 countries as features, the covariance matrix will be $16 \times 16$.")
print("Again, we use `np.cov` with `rowvar=False` as features are in columns (countries).")

covariance_matrix_food_items = np.cov(data_centered_food_items, rowvar=False)

print("\nShape of the covariance matrix for food items PCA:", covariance_matrix_food_items.shape)
print("First 3x3 block of the covariance matrix:")
print(covariance_matrix_food_items[:3, :3])
```

```python
# Cell 11: Part 2 - Eigenvalues and Eigenvectors for Food Items PCA
print("#### Computing Eigenvalues and Eigenvectors (Food Items PCA)")
print("We compute the eigenvalues and eigenvectors of this new covariance matrix.")
print("The eigenvalues will tell us how much variance is explained by each principal component in this transposed space, and the eigenvectors will represent combinations of country consumption patterns.")

eigenvalues_food_items, eigenvectors_food_items = np.linalg.eigh(covariance_matrix_food_items)

# Sort eigenvalues and eigenvectors in descending order
sorted_indices_food_items = np.argsort(eigenvalues_food_items)[::-1]
eigenvalues_food_items_sorted = eigenvalues_food_items[sorted_indices_food_items]
eigenvectors_food_items_sorted = eigenvectors_food_items[:, sorted_indices_food_items]

print("\nSorted Eigenvalues (top 5 for food items PCA):")
print(eigenvalues_food_items_sorted[:5])
print("\nThese eigenvalues indicate the importance of each principal component in explaining the variance among food items based on country consumption.")

print("\nFirst two principal components (eigenvectors for food items PCA):")
principal_components_food_items = eigenvectors_food_items_sorted[:, :2]
print(principal_components_food_items)
print("Each column represents a principal component, which is a linear combination of country consumption patterns.")
```

```python
# Cell 12: Part 2 - Projecting Data and Scatter Plot for Food Items PCA
print("#### Projecting Data onto Principal Components and Scatter Plot (Food Items PCA)")
print("Finally, we project the mean-centered food item data onto the first two principal components derived from country consumption patterns.")
print("This will give us a 2-dimensional representation of each food item, allowing us to visualize relationships between them.")

projected_data_food_items = np.dot(data_centered_food_items, principal_components_food_items)

# Create a DataFrame for the projected data with food item names
projected_df_food_items = pd.DataFrame(projected_data_food_items,
                                       columns=['PC1', 'PC2'],
                                       index=df_pca_food_items.index)

print("\nProjected data for food items (first 5 rows):")
print(projected_df_food_items.head())

print("\nWe will now generate a scatter plot to visualize the food items in this 2D principal component space.")
plt.figure(figsize=(12, 9))
plt.scatter(projected_df_food_items['PC1'], projected_df_food_items['PC2'], s=100, alpha=0.7)

# Annotate each point with the food item name
for i, food_item in enumerate(projected_df_food_items.index):
    plt.annotate(food_item, (projected_df_food_items['PC1'][i], projected_df_food_items['PC2'][i]),
                 textcoords="offset points", xytext=(5,5), ha='center')

plt.title('PCA of Food Items based on Country Consumption')
plt.xlabel('Principal Component 1 (PC1)')
plt.ylabel('Principal Component 2 (PC2)')
plt.grid(True)
plt.axhline(0, color='grey', linewidth=0.8)
plt.axvline(0, color='grey', linewidth=0.8)
plt.show() # Display the plot in the notebook
```

```python
# Cell 13: Part 2 - Analysis of the Scatter Plot for Food Items PCA
print("#### Analysis of the Scatter Plot (Food Items PCA)")
print("The scatter plot of food items, based on their consumption patterns across countries, reveals interesting groupings and relationships:")
print("- **Categorization of Food Items:** Food items that are consumed similarly across countries tend to group together.")
print("  - For example, 'Real coffee', 'Tea', and 'Instant coffee' are relatively close, suggesting that countries that consume one tend to consume the others similarly.")
print("  - 'Sweetener', 'Crisp bread', and 'Margarine' also appear to form a cluster, potentially indicating a dietary preference or lifestyle choice across countries.")
print("  - 'Olive oil' and 'Garlic' are often found together, which makes sense considering their common use in Mediterranean diets.")
print("- **Opposing Consumption Trends:** Food items that are far apart on the plot indicate contrasting consumption patterns across countries.")
print("  - For instance, 'Tin soup' and 'Powder soup' are often clustered together, possibly indicating a convenience food category.")
print("  - The spread along PC1 and PC2 suggests different 'dietary axes'. One axis might separate traditional ingredients from convenience foods, while another might distinguish between animal products and plant-based foods, or specific regional preferences.")
print("- **Interpretation of Axes (PC1 and PC2):**")
print("  - **PC1 (Horizontal Axis):** This component likely represents a major dietary difference among countries. For example, it might distinguish between countries with high consumption of certain staples vs. others.")
print("  - **PC2 (Vertical Axis):** This component captures the second most significant variation. It could represent a secondary dietary trend, such as a preference for fresh produce vs. processed foods.")
print("This visualization helps in understanding which food items are frequently consumed together or show similar consumption trends across the diverse European countries, providing insights into various dietary habits.")
```
